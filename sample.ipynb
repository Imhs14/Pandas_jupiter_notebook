Introduction to Pandas

What is Pandas ?

- A powerful library for data manipulation and analysis in python built on top of Numpy. Provides data structure like series and DataFrame

Why Use Pandas?

- Easy data Cleaning and Transformation.
- Efficient handling of large datasets.
- integration with other Python libraries like matplotlib and seaborn for visualization

```python
pd.merge(df1,df2, on='name') # The pandas will ignore the stuff that is not there in anyone of the df1, df2 it will only consider the stuff which is there in both of the dataframes.
```
